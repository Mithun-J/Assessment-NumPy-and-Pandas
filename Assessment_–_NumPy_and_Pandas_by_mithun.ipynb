```python
import pandas as pd
import numpy as np

student_performance = pd.read_csv("/content/drive/MyDrive/DSAB13/Colab Notebooks/Dataset/Assesement dataset/student_performance.csv")

# question 11

print("First 5 records:")
print(student_performance.head())
print("")

print("Last 5 records:")
print(student_performance.tail())
print("")

print("Number of rows and columns:")
print(student_performance.shape)
print("")

print("Column names:")
print(student_performance.columns)
print("")

print("Data types:")
print(student_performance.dtypes)
print("")

print("Statistical summary:")
print(student_performance.describe())
```

```python
# question 12

print("Students with Final_Score greater than 75:")
print(student_performance[student_performance["Final_Score"] > 75])
print("")

print("Students with Attendance below 75%:")
print(student_performance[student_performance["Attendance"] < 75])
print("")

print("Data Science students with Final_Score greater than 80:")
print(student_performance[(student_performance["Department"] == "Data Science") &(student_performance["Final_Score"] > 80)])
print("")

print("Students sorted by Final_Score:")
print(student_performance.sort_values(by="Final_Score",ascending=False)
)
```

```python
# question 13

student_performance["Average_Technical_Score"] = student_performance[["Python_Score", "SQL_Score", "Pandas_Score", "NumPy_Score"]].mean(axis=1)

student_performance["Performance"] = np.select([
        student_performance["Average_Technical_Score"] >= 90,
        student_performance["Average_Technical_Score"] >= 75,
        student_performance["Average_Technical_Score"] >= 60],
    ["Excellent","Good","Average"],default="Needs Improvement")

display(student_performance)
```

```python
# question 14

department_result = pd.DataFrame({
    "Number_of_Students": student_performance.groupby("Department")["Student_ID"].count(),
    "Average_Final_Score": student_performance.groupby("Department")["Final_Score"].mean(),
    "Maximum_Final_Score": student_performance.groupby("Department")["Final_Score"].max(),
    "Minimum_Final_Score": student_performance.groupby("Department")["Final_Score"].min()
})

department_result = department_result.sort_values("Average_Final_Score",ascending=False)
display(department_result)
```

```python
# question 15

scores = student_performance["Final_Score"].dropna().values

print("Mean:", scores.mean())
print("Median:", np.median(scores))
print("Standard Deviation:", scores.std())
print("25th Percentile:", np.percentile(scores, 25))
print("75th Percentile:", np.percentile(scores, 75))

print("Students with Final_Score above the overall mean:")

display(student_performance[student_performance["Final_Score"] > scores.mean()]
)
```

```python
# question 16

# Load the dataset
student_performance = pd.read_csv("/content/drive/MyDrive/DSAB13/Colab Notebooks/Dataset/Assesement dataset/student_performance.csv")

# Part A - Data Cleaning
print("Missing Values:")
print(student_performance.isnull().sum())

student_performance["Final_Score"] = student_performance["Final_Score"].fillna(student_performance["Final_Score"].mean())

student_performance["Attendance"] = student_performance["Attendance"].fillna(student_performance["Attendance"].mean())

print("Missing Values After Cleaning:")
print(student_performance.isnull().sum())
print("")


# Part B - Feature Engineering

student_performance["Average_Score"] = student_performance[["Python_Score", "SQL_Score", "Pandas_Score", "NumPy_Score"]].mean(axis=1)

student_performance["Pass_Status"] = np.where(student_performance["Average_Score"] >= 50,"Pass","Fail")

print("Student Performance After Feature Engineering:")
display(student_performance)
print("")

# Part C - Analysis

# Top 5 students by Average_Score
print("Top 5 Students:")
top_5_students = student_performance.sort_values("Average_Score", ascending=False).head(5)
display(top_5_students)
print("")


# Department with highest average score
print("Average Score by Department:")
department_average = student_performance.groupby("Department")["Average_Score"].mean()
print(department_average)

print("Department with Highest Average Score:")
print(department_average.idxmax())
print("")


# Average attendance by department
print("Average Attendance by Department:")
attendance_by_department = student_performance.groupby("Department")["Attendance"].mean()
print(attendance_by_department)
print("")

# Number of students passed and failed
print("Number of Students Passed:")
print((student_performance["Pass_Status"] == "Pass").sum())

print("Number of Students Failed:")
print((student_performance["Pass_Status"] == "Fail").sum())
print("")

# Part D - Summary

print("Summary:")

summary = pd.DataFrame({
    "Total_Students": student_performance.groupby("Department")["Student_ID"].count(),
    "Average_Score": student_performance.groupby("Department")["Average_Score"].mean(),
    "Average_Attendance": student_performance.groupby("Department")["Attendance"].mean(),
    "Pass_Count": student_performance[student_performance["Pass_Status"] == "Pass"].groupby("Department")["Student_ID"].count(),
    "Fail_Count": student_performance[student_performance["Pass_Status"] == "Fail"].groupby("Department")["Student_ID"].count()
})

summary = summary.fillna(0)

summary = summary.reset_index()

display(summary)
```
